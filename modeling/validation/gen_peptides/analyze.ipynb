```python
import pandas as pd
import numpy as np

# there are 3 files, generated_fibers_init.txt, filtered_ap_peptides_fiber.txt and valid_fibers.csv
# these all correspond to the same set of fibers, but only with the ones passing each filtering step
# we want to know which fibers did not pass each filtering step

np.random.seed(42)

fibers_init = pd.read_csv('generated_fibers_init.txt', header=None, names=['fiber_id'])
fibers_init = fibers_init.drop_duplicates()
fibers_filtered_ap = pd.read_csv('filtered_ap_peptides_fiber.txt', header=None, names=['fiber_id', 'fiber_id_3', 'ap', 'agg_prob'])
fibers_valid = pd.read_csv('valid_fibers.csv', header=0, names=['fiber_id'])
```

```python
# find fibers that did not pass the AP peptide filtering step
fibers_not_passing_ap = fibers_init[~fibers_init['fiber_id'].isin(fibers_filtered_ap['fiber_id'])]
fibers_not_passing_ap
```

```python
# find fibers that did not pass the final validation step
fibers_not_passing_valid = fibers_filtered_ap[~fibers_filtered_ap['fiber_id'].isin(fibers_valid['fiber_id'])]
fibers_not_passing_valid
```

```python
len(fibers_init) - len(fibers_not_passing_ap) - len(fibers_not_passing_valid) == len(fibers_valid)
```

```python
fibers_valid
```

```python
# do the same for the spheres
spheres_init = pd.read_csv('generated_spheres_init.txt', header=None, names=['sphere_id'])
spheres_init = spheres_init.drop_duplicates()
filtered_ap_spheres = pd.read_csv('filtered_ap_peptides_spheres.txt', header=None, names=['sphere_id', 'sphere_id_3', 'ap', 'agg_prob'])
spheres_valid = pd.read_csv('valid_spheres.csv', header=0, names=['sphere_id'])
```

```python
spheres_not_passing_ap = spheres_init[~spheres_init['sphere_id'].isin(filtered_ap_spheres['sphere_id'])]
spheres_not_passing_ap
```

```python
spheres_not_passing_valid = filtered_ap_spheres[~filtered_ap_spheres['sphere_id'].isin(spheres_valid['sphere_id'])]
spheres_not_passing_valid
```

```python
len(spheres_init) - len(spheres_not_passing_ap) - len(spheres_not_passing_valid) == len(spheres_valid)
```

```python
spheres_valid
```

```python
fibers_neg_control_ap = fibers_not_passing_ap.sample(n=15, random_state=42)
spheres_neg_control_ap = spheres_not_passing_ap.sample(n=15, random_state=42)
fibers_neg_control_valid = fibers_not_passing_valid.sample(n=15, random_state=42)
spheres_neg_control_valid = spheres_not_passing_valid.sample(n=15, random_state=42)

# merge into a single dataframe, instead of 'fiber_id' or 'sphere_id' use 'structure_id' and add a column 'structure_type' with values 'fiber' or 'sphere'
neg_control_ap = pd.concat([
    fibers_neg_control_ap.rename(columns={'fiber_id': 'structure_id', 'fiber_id_3': 'structure_id_3'}).assign(structure_type='fiber'),
    spheres_neg_control_ap.rename(columns={'sphere_id': 'structure_id', 'sphere_id_3': 'structure_id_3'}).assign(structure_type='sphere')
])

neg_control_valid = pd.concat([
    fibers_neg_control_valid.rename(columns={'fiber_id': 'structure_id', 'fiber_id_3': 'structure_id_3'}).assign(structure_type='fiber'),
    spheres_neg_control_valid.rename(columns={'sphere_id': 'structure_id', 'sphere_id_3': 'structure_id_3'}).assign(structure_type='sphere')
])

# merge both neg control dataframes
neg_control = pd.concat([neg_control_ap, neg_control_valid])
neg_control.reset_index(drop=True, inplace=True)
neg_control
```

```python
# count how many sequences with len < 5 are in the neg_control, and for which type
# if it's NaN in ap or agg_prob, then it did not pass the ap filtering step
# if it's not, it did not pass the final validation step
# check how many are from each step
neg_control['sequence_length'] = neg_control['structure_id'].apply(len)
neg_control_short_seqs_ap = neg_control[(neg_control['sequence_length'] < 5) & (neg_control['ap'].isna())]
neg_control_short_seqs_valid = neg_control[(neg_control['sequence_length'] < 5) & (neg_control['ap'].notna())]

neg_control_short_seqs_ap, neg_control_short_seqs_valid

```

```python
# sample structures with len >= 5 as substitutes for these len < 5 sequences
# from spheres_not_passing_ap if ap is NaN, else from spheres_not_passing_valid
neg_control_ap_substitutes = spheres_not_passing_ap[spheres_not_passing_ap['sphere_id'].apply(len) >= 5].sample(n=len(neg_control_short_seqs_ap), random_state=42)
neg_control_valid_substitutes = spheres_not_passing_valid[spheres_not_passing_valid['sphere_id'].apply(len) >= 5].sample(n=len(neg_control_short_seqs_valid), random_state=42)
# add missing columns to match neg_control
neg_control_ap_substitutes, neg_control_valid_substitutes
```

```python
neg_control_short_seqs_valid
```

```python
# add these substitutes to the neg_control dataframe, replacing the short sequences
neg_control_final = neg_control.copy()

# drop from neg_control_final the short sequences
neg_control_final = neg_control_final[~((neg_control_final['sequence_length'] < 5))]

# add the substitutes
neg_control_ap_substitutes = neg_control_ap_substitutes.rename(columns={'sphere_id': 'structure_id', 'sphere_id_3': 'structure_id_3'})
neg_control_ap_substitutes = neg_control_ap_substitutes.assign(structure_type='sphere', ap=np.nan, agg_prob=np.nan, sequence_length=neg_control_ap_substitutes['structure_id'].apply(len))
neg_control_valid_substitutes = neg_control_valid_substitutes.rename(columns={'sphere_id': 'structure_id', 'sphere_id_3': 'structure_id_3'})
neg_control_valid_substitutes = neg_control_valid_substitutes.assign(structure_type='sphere', sequence_length=neg_control_valid_substitutes['structure_id'].apply(len))
neg_control_final = pd.concat([neg_control_final, neg_control_ap_substitutes, neg_control_valid_substitutes])
neg_control_final
```

```python
# check per length of structure_id
neg_control_final['structure_length'] = neg_control_final['structure_id'].apply(len)

# do a count per structure_type and structure_length
neg_control_counts = neg_control_final.groupby(['structure_length']).size().reset_index(name='count')
neg_control_counts
```

```python
all_lengths = range(5, 11)
ALPHABET = 'ACDEFGHIKLMNPQRSTVWY'
np.random.seed(42)
current_count = 12

def generate_random_peptide(length):
    return ''.join(np.random.choice(list(ALPHABET), size=length))

new_peptides = []
while current_count < 15:
    length = np.random.choice(all_lengths)
    random_peptide_1= generate_random_peptide(length)
    neg_control_final = pd.concat([neg_control_final, pd.DataFrame({
        'structure_id': [random_peptide_1],
        'structure_id_3': [None],
        'structure_type': ['random'],
        'structure_length': [length]
    })], ignore_index=True)
    current_count += 1
    new_peptides.append(random_peptide_1)

neg_control_final.reset_index(drop=True, inplace=True)
neg_control_final
```

```python
# save to csv
neg_control_final.to_csv('negative_control_structures_new.csv', index=False)

# save peptides as a fasta file
with open('negative_control_structures_new.fasta', 'w') as fasta_file:
    for idx, row in neg_control_final.iterrows():
        fasta_file.write(f">{row['structure_type']}_{idx}\n")
        fasta_file.write(f"{row['structure_id']}\n")
```
