```python
# load original csv
original = pd.read_csv('negative_control_structures.csv')
# rename structure_id to peptide
original = original.rename(columns={'structure_id': 'peptide'})
original
```

```python
# create a column 'class'
# if structure_type='random', then class='random'
# if structure_type='unsupervised', then class='unsupervised'
# if structure_type='fiber' or 'sphere' and ap=NaN, then class='fiber_no_ap' or 'sphere_no_ap'
# if structure_type='fiber' or 'sphere' and ap!=NaN, then class='fiber_no_desc' or 'sphere_no_desc'

def classify(row):
    if row['structure_type'] == 'random':
        return 'random'
    elif row['structure_type'] == 'unsupervised':
        return 'unsupervised'
    elif row['structure_type'] == 'fiber':
        if pd.isna(row['ap']):
            return 'fiber_no_ap'
        else:
            return 'fiber_no_desc'
    elif row['structure_type'] == 'sphere':
        if pd.isna(row['ap']):
            return 'sphere_no_ap'
        else:
            return 'sphere_no_desc'
    else:
        return 'unknown'
original['class'] = original.apply(classify, axis=1)
original
```

```python
# adapt merged to have a 'has_aggregated' column
# also a 'matches_morphology' for fiber and sphere targets, if RMOI_mean is >=0.75 for spheres and <=0.35 for fibers
def has_aggregated(row):
    return row['aggregation_propensity_mean'] >= 1.8

def matches_morphology(row):
    if not has_aggregated(row):
        return False
    if row['class'] == 'fiber_no_desc' or row['class'] == 'fiber_no_ap' or row['class'] == 'fiber_pepmorph':
        return row['RMOI_mean'] <= 0.35
    elif row['class'] == 'sphere_no_desc' or row['class'] == 'sphere_no_ap' or row['class'] == 'sphere_pepmorph':
        return row['RMOI_mean'] >= 0.75
    else:
        return False
    
merged['has_aggregated'] = merged.apply(has_aggregated, axis=1)
merged['matches_morphology'] = merged.apply(matches_morphology, axis=1)
merged
```

```python
# apply has_aggregated and matches_morphology by run
def has_aggregated_by_run(row):
    return row['aggregation_propensity'] >= 1.8

def matches_morphology_by_run(row):
    if not has_aggregated_by_run(row):
        return False
    if row['class'] == 'fiber_no_desc' or row['class'] == 'fiber_no_ap' or row['class'] == 'fiber_pepmorph':
        return row['RMOI'] <= 0.35
    elif row['class'] == 'sphere_no_desc' or row['class'] == 'sphere_no_ap' or row['class'] == 'sphere_pepmorph':
        return row['RMOI'] >= 0.75
    else:
        return False

def matches_morphology_visually_by_run(row):
    return row['is_target'] == "yes"

merged_by_run['has_aggregated'] = merged_by_run.apply(has_aggregated_by_run, axis=1)
merged_by_run['matches_morphology'] = merged_by_run.apply(matches_morphology_by_run, axis=1)
merged_by_run['matches_morphology_visually'] = merged_by_run.apply(matches_morphology_visually_by_run, axis=1)
merged_by_run
```

```python
import numpy as np
import pandas as pd
from collections import Counter
from scipy.stats import fisher_exact
from statsmodels.stats.contingency_tables import Table2x2
```

```python
def wilson_ci(k: int, n: int, alpha: float = 0.05):
    """Wilson score CI for a binomial proportion."""
    if n == 0:
        return (np.nan, np.nan)
    z = 1.959963984540054  # ~N(0,1) 97.5% quantile for 95% CI
    p = k / n
    denom = 1 + z*z/n
    center = (p + z*z/(2*n)) / denom
    half = (z * np.sqrt((p*(1-p)/n) + (z*z/(4*n*n)))) / denom
    return (float(center - half), float(center + half))

def fisher_or_ci(a, b, c, d, alpha: float = 0.05):
    """
    2x2 with:
      PepMorph: success=a, fail=b
      Control : success=c, fail=d
    Returns odds_ratio, (lo, hi), p_value.
    """
    table_int = np.array([[a, b], [c, d]], dtype=int)
    or_scipy, p = fisher_exact(table_int, alternative="two-sided")

    # CI for OR
    if (table_int == 0).any():
        # Haldane-Anscombe correction for stability
        table = table_int.astype(float) + 0.5
        ct = Table2x2(table)
        or_hat = float(ct.oddsratio)
        lo, hi = ct.oddsratio_confint(alpha=alpha, method="wald")
    else:
        ct = Table2x2(table_int)
        or_hat = float(ct.oddsratio)
        try:
            lo, hi = ct.oddsratio_confint(alpha=alpha, method="exact")
        except Exception:
            lo, hi = ct.oddsratio_confint(alpha=alpha, method="wald")
    return or_hat, (float(lo), float(hi)), float(p)

```

```python
import pandas as pd

def normalize_visual_label(row):
    """
    Per-run label used ONLY for voting:
      'S' = sphere-like
      'F' = fibril/tube-like
      'O' = other / not classified as S or F / did not match target
    """
    cls = str(row.get("class", "")).strip().lower()
    it  = row.get("is_target", None)
    it  = "" if (it is None or (isinstance(it, float) and pd.isna(it))) else str(it).strip().lower()

    if it in {"t", "tube", "tubes", "tubular", "fibril", "fibrils"}:
        it = "fiber"
    if it in {"s", "sphere", "spheres", "spherical"}:
        it = "sphere"

    if cls in {"random", "unsupervised"}:
        if it == "sphere":
            return "S"
        if it == "fiber":
            return "F"
        return "O"

    target = None
    if "sphere" in cls:
        target = "S"
    elif ("fiber" in cls) or ("fibril" in cls):
        target = "F"

    if it == "yes":
        return target if target is not None else "O"
    if it == "no" or it == "":
        return "O"

    if it == "sphere":
        return "S"
    if it == "fiber":
        return "F"

    return "O"

def majority_vote_strict(vals):
    """
    Returns:
      'S' or 'F' only if it has a strict majority among votes (including 'O').
      Otherwise returns 'O'.

    With 3 runs: requires 2 votes for S or F.
    """
    vals = [v for v in vals if v is not None]
    if not vals:
        return None

    c = Counter(vals)
    n = len(vals)
    need = (n // 2) + 1

    if c["S"] >= need:
        return "S"
    if c["F"] >= need:
        return "F"
    return "O"


def collapse_runs_to_peptides(df_runs: pd.DataFrame,
                              ap_thresh=1.8,
                              rmoi_sphere_thresh=0.75,
                              rmoi_fiber_thresh=0.35):
    df = df_runs.copy()

    required = ["peptide", "class", "run", "aggregation_propensity", "RMOI", "is_target"]
    missing = [c for c in required if c not in df.columns]
    if missing:
        raise ValueError(f"CSV missing required columns: {missing}")

    df["aggregation_propensity"] = pd.to_numeric(df["aggregation_propensity"], errors="coerce")
    df["RMOI"] = pd.to_numeric(df["RMOI"], errors="coerce")

    df["ap_pass"] = df["aggregation_propensity"] >= ap_thresh
    df["rmoi_pass_sphere"] = df["RMOI"] >= rmoi_sphere_thresh
    df["rmoi_pass_fiber"]  = df["RMOI"] <= rmoi_fiber_thresh

    df["vis_norm"] = df.apply(normalize_visual_label, axis=1)

    def strict_majority_bool(bools):
        bools = [bool(x) for x in bools if x is not None and not (isinstance(x, float) and pd.isna(x))]
        if not bools:
            return False
        need = (len(bools) // 2) + 1
        return sum(bools) >= need

    g = df.groupby(["class", "peptide"], as_index=False).agg(
        ap_mean=("aggregation_propensity", "mean"),
        rmoi_mean=("RMOI", "mean"),
        ap_majority=("ap_pass", strict_majority_bool),
        rmoi_majority_sphere=("rmoi_pass_sphere", strict_majority_bool),
        rmoi_majority_fiber=("rmoi_pass_fiber", strict_majority_bool),
        vis_majority=("vis_norm", lambda s: majority_vote_strict(list(s))),
        vis_any_S=("vis_norm", lambda s: any(v == "S" for v in s if v is not None)),
        vis_any_F=("vis_norm", lambda s: any(v == "F" for v in s if v is not None)),
    )
    return g
```

```python
def compute_descriptive_values(
    df_pep: pd.DataFrame,
    *,
    ap_thresh: float = 1.8,
    rmoi_sphere_thresh: float = 0.75,
    rmoi_fiber_thresh: float = 0.35,
    class_meta: dict,
    untargeted_mode: str = "majority",
):
    """
    Returns a DataFrame with all values needed for the paper Table 6.

    class_meta: dict mapping class -> dict with keys:
      - section: "Untargeted" / "Spherical-targeted" / "Fibril-targeted"
      - cohort: row label, e.g. "Failed AP screen"
      - target: None (untargeted) or "sphere" or "fiber"
    """
    df = df_pep.copy()

    df["has_agg"] = df["ap_majority"]
    df["rmoi_is_sphere"] = df["rmoi_majority_sphere"]
    df["rmoi_is_fiber"]  = df["rmoi_majority_fiber"]

    rows = []
    for cls, meta in class_meta.items():
        sub = df[df["class"] == cls].copy()
        n = int(sub["peptide"].nunique())
        agg = sub[sub["has_agg"]].copy()
        n_agg = int(agg["peptide"].nunique())

        # Aggregation rate (unconditional)
        agg_rate = (n_agg / n) if n else np.nan

        target = meta.get("target", None)  # None / "sphere" / "fiber"

        out = dict(
            class_key=cls,
            section=meta.get("section", ""),
            cohort=meta.get("cohort", cls),
            n_total=n,
            n_agg=n_agg,
            agg_rate=agg_rate,
        )

        if target is None:
            # Untargeted: report S and F fractions among aggregators
            if n_agg == 0:
                out.update(
                    rmoi_S_k=0, rmoi_F_k=0, vis_S_k=0, vis_F_k=0,
                    rmoi_S_rate=np.nan, rmoi_F_rate=np.nan,
                    vis_S_rate=np.nan, vis_F_rate=np.nan,
                )
                rows.append(out)
                continue

            if untargeted_mode == "any_run":
                rmoi_S_k = int(agg["rmoi_any_S"].sum())
                rmoi_F_k = int(agg["rmoi_any_F"].sum())
                vis_S_k  = int(agg["vis_any_S"].sum())
                vis_F_k  = int(agg["vis_any_F"].sum())
            elif untargeted_mode == "majority":
                rmoi_S_k = int(agg["rmoi_is_sphere"].sum())
                rmoi_F_k = int(agg["rmoi_is_fiber"].sum())
                vis_S_k  = int((agg["vis_majority"] == "S").sum())
                vis_F_k  = int((agg["vis_majority"] == "F").sum())
            else:
                raise ValueError("untargeted_mode must be 'majority' or 'any_run'")

            out.update(
                rmoi_S_k=rmoi_S_k,
                rmoi_F_k=rmoi_F_k,
                vis_S_k=vis_S_k,
                vis_F_k=vis_F_k,
                rmoi_S_rate=rmoi_S_k / n_agg,
                rmoi_F_rate=rmoi_F_k / n_agg,
                vis_S_rate=vis_S_k / n_agg,
                vis_F_rate=vis_F_k / n_agg,
            )
            rows.append(out)
            continue

        # Targeted cohorts: success conditional on aggregation
        if n_agg == 0:
            out.update(
                rmoi_succ_k=0, rmoi_succ_rate=np.nan,
                vis_succ_k=0,  vis_succ_rate=np.nan,
            )
            rows.append(out)
            continue

        if target == "sphere":
            rmoi_succ_k = int(agg["rmoi_is_sphere"].sum())
            vis_succ_k  = int((agg["vis_majority"] == "S").sum())
        elif target == "fiber":
            rmoi_succ_k = int(agg["rmoi_is_fiber"].sum())
            vis_succ_k  = int((agg["vis_majority"] == "F").sum())
        else:
            raise ValueError(f"Unknown target: {target}")

        out.update(
            rmoi_succ_k=rmoi_succ_k,
            rmoi_succ_rate=rmoi_succ_k / n_agg,
            vis_succ_k=vis_succ_k,
            vis_succ_rate=vis_succ_k / n_agg,
        )
        rows.append(out)

    out_df = pd.DataFrame(rows)

    # Convenience formatted fields matching the table style
    def fmt_pct(k, n):
        return "--" if n == 0 else f"{100.0*k/n:.1f}% ({k}/{n})"

    out_df["agg_rate_str"] = out_df.apply(lambda r: fmt_pct(int(r["n_agg"]), int(r["n_total"])), axis=1)

    # For untargeted vs targeted, add corresponding strings
    def rmoi_str(r):
        if r["section"] == "Untargeted":
            if r["n_agg"] == 0:
                return "S: -- ; F: --"
            return f"S: {fmt_pct(int(r['rmoi_S_k']), int(r['n_agg']))} ; F: {fmt_pct(int(r['rmoi_F_k']), int(r['n_agg']))}"
        else:
            if r["n_agg"] == 0:
                return "--"
            return fmt_pct(int(r["rmoi_succ_k"]), int(r["n_agg"]))

    def vis_str(r):
        if r["section"] == "Untargeted":
            if r["n_agg"] == 0:
                return "S: -- ; F: --"
            return f"S: {fmt_pct(int(r['vis_S_k']), int(r['n_agg']))} ; F: {fmt_pct(int(r['vis_F_k']), int(r['n_agg']))}"
        else:
            if r["n_agg"] == 0:
                return "--"
            return fmt_pct(int(r["vis_succ_k"]), int(r["n_agg"]))

    out_df["rmoi_outcome_str"] = out_df.apply(rmoi_str, axis=1)
    out_df["visual_outcome_str"] = out_df.apply(vis_str, axis=1)

    return out_df


def compute_rates_values(
    df_pep: pd.DataFrame,
    *,
    ap_thresh: float = 1.8,
    sphere_pepmorph_class: str,
    sphere_descfail_class: str,
    fiber_pepmorph_class: str,
    fiber_descfail_class: str,
):
    """
    Computes visual outcomes (conditional on aggregation) and Fisher OR/p for:
      - Spheres: PepMorph vs Desc-fail
      - Fibrils: PepMorph vs Desc-fail
      - All pooled: combine both targets
    Returns (rates_df, stats_df).
    """
    df = df_pep.copy()
    df["has_agg"] = df["ap_majority"]

    def block(target_name, pep_cls, ctrl_cls, success_label):
        pep = df[(df["class"] == pep_cls) & (df["has_agg"])].copy()
        ctrl = df[(df["class"] == ctrl_cls) & (df["has_agg"])].copy()

        n1 = int(pep["peptide"].nunique())
        n0 = int(ctrl["peptide"].nunique())

        s1 = int((pep["vis_majority"] == success_label).sum())
        s0 = int((ctrl["vis_majority"] == success_label).sum())

        lo1, hi1 = wilson_ci(s1, n1)
        lo0, hi0 = wilson_ci(s0, n0)

        or_hat, (or_lo, or_hi), p = fisher_or_ci(s1, n1 - s1, s0, n0 - s0)

        rates = [
            dict(target=target_name, cohort="PepMorph", success_k=s1, n=n1, rate=(s1/n1 if n1 else np.nan), ci_lo=lo1, ci_hi=hi1),
            dict(target=target_name, cohort="Desc.-fail", success_k=s0, n=n0, rate=(s0/n0 if n0 else np.nan), ci_lo=lo0, ci_hi=hi0),
        ]
        stats = dict(target=target_name, fisher_or=or_hat, or_ci_lo=or_lo, or_ci_hi=or_hi, p_value=p,
                     pep_success=s1, pep_n=n1, ctrl_success=s0, ctrl_n=n0)
        return rates, stats

    rates_all = []
    stats_all = []

    r, s = block("Spheres", sphere_pepmorph_class, sphere_descfail_class, success_label="S")
    rates_all += r; stats_all.append(s)

    r, s = block("Fibrils", fiber_pepmorph_class, fiber_descfail_class, success_label="F")
    rates_all += r; stats_all.append(s)

    # pooled
    # pool the underlying 2x2 counts directly
    sph = stats_all[0]
    fib = stats_all[1]
    a = sph["pep_success"] + fib["pep_success"]
    b = (sph["pep_n"] - sph["pep_success"]) + (fib["pep_n"] - fib["pep_success"])
    c = sph["ctrl_success"] + fib["ctrl_success"]
    d = (sph["ctrl_n"] - sph["ctrl_success"]) + (fib["ctrl_n"] - fib["ctrl_success"])

    # pooled rates/CI
    pep_n = sph["pep_n"] + fib["pep_n"]
    pep_s = sph["pep_success"] + fib["pep_success"]
    ctrl_n = sph["ctrl_n"] + fib["ctrl_n"]
    ctrl_s = sph["ctrl_success"] + fib["ctrl_success"]

    pep_lo, pep_hi = wilson_ci(pep_s, pep_n)
    ctrl_lo, ctrl_hi = wilson_ci(ctrl_s, ctrl_n)

    pooled_or, (pooled_lo, pooled_hi), pooled_p = fisher_or_ci(a, b, c, d)

    rates_all += [
        dict(target="All pooled", cohort="PepMorph", success_k=pep_s, n=pep_n, rate=(pep_s/pep_n if pep_n else np.nan), ci_lo=pep_lo, ci_hi=pep_hi),
        dict(target="All pooled", cohort="Desc.-fail", success_k=ctrl_s, n=ctrl_n, rate=(ctrl_s/ctrl_n if ctrl_n else np.nan), ci_lo=ctrl_lo, ci_hi=ctrl_hi),
    ]
    stats_all.append(dict(target="All pooled", fisher_or=pooled_or, or_ci_lo=pooled_lo, or_ci_hi=pooled_hi, p_value=pooled_p,
                          pep_success=pep_s, pep_n=pep_n, ctrl_success=ctrl_s, ctrl_n=ctrl_n))

    rates_df = pd.DataFrame(rates_all)
    stats_df = pd.DataFrame(stats_all)

    return rates_df, stats_df
```

```python
CSV_PATH = "analysis_by_run.csv"
df_runs = pd.read_csv(CSV_PATH)
df_pep = collapse_runs_to_peptides(df_runs)

class_meta = {
    "random":          {"section": "Untargeted",          "cohort": "Random",                 "target": None},
    "unsupervised":    {"section": "Untargeted",          "cohort": "Unsupervised",           "target": None},

    "sphere_no_ap":    {"section": "Spherical-targeted",  "cohort": "Failed AP screen",       "target": "sphere"},
    "sphere_no_desc":  {"section": "Spherical-targeted",  "cohort": "Failed descriptor screen","target": "sphere"},
    "sphere_pepmorph": {"section": "Spherical-targeted",  "cohort": "PepMorph candidates",    "target": "sphere"},

    "fiber_no_ap":     {"section": "Fibril-targeted",     "cohort": "Failed AP screen",       "target": "fiber"},
    "fiber_no_desc":   {"section": "Fibril-targeted",     "cohort": "Failed descriptor screen","target": "fiber"},
    "fiber_pepmorph":  {"section": "Fibril-targeted",     "cohort": "PepMorph candidates",    "target": "fiber"},
}

tab = compute_descriptive_values(
    df_pep,
    ap_thresh=1.8,
    rmoi_sphere_thresh=0.75,
    rmoi_fiber_thresh=0.35,
    class_meta=class_meta,
    untargeted_mode="majority",  # or "any_run" if you want S/F to potentially overlap
)

rates, stats = compute_rates_values(
    df_pep,
    ap_thresh=1.8,
    sphere_pepmorph_class="sphere_pepmorph",
    sphere_descfail_class="sphere_no_desc",
    fiber_pepmorph_class="fiber_pepmorph",
    fiber_descfail_class="fiber_no_desc",
)
```

```python
BB_POINT_SIZE = 8
SC_POINT_SIZE = 6

CUTOFF_NM = 1.8
MIN_PNG_BYTES = 2000

USE_FILESYSTEM = False

_PNG_MAGIC = b"\x89PNG\r\n\x1a\n"

def is_valid_png(path: Path, min_bytes=MIN_PNG_BYTES):
    try:
        path = Path(path)
        if not path.exists(): return False
        if path.stat().st_size < min_bytes: return False
        with path.open("rb") as f:
            return f.read(8) == _PNG_MAGIC
    except Exception:
        return False


def _residue_anchors(U):
    """One anchor per residue (BB if present; else residue COM). Units: Å."""
    anchors = []
    for res in U.residues:
        bb = res.atoms.select_atoms("name BB")
        anchors.append(bb.positions[0] if bb.n_atoms else res.atoms.center_of_mass())
    return np.asarray(anchors, dtype=float)

def _largest_residue_cluster_from_anchors(anchors_A, cutoff_nm, box):
    """
    anchors_A: (n,3) Å
    cutoff_nm: nm
    box: MDAnalysis ts.dimensions (Å + angles)
    """
    cutoff_A = cutoff_nm * 10.0
    pairs = capped_distance(
        anchors_A, anchors_A, cutoff_A,
        box=box, return_distances=False
    )

    n = len(anchors_A)
    parent = np.arange(n, dtype=int)

    def find(x):
        while parent[x] != x:
            parent[x] = parent[parent[x]]
            x = parent[x]
        return x

    def union(a, b):
        ra, rb = find(a), find(b)
        if ra != rb:
            parent[rb] = ra

    for i, j in pairs:
        if i != j:
            union(int(i), int(j))

    comps = {}
    for i in range(n):
        r = find(i)
        comps.setdefault(r, []).append(i)

    largest = max(comps.values(), key=len) if comps else [0]
    return np.array(largest, dtype=int)

def center_main_cluster_inplace_on_current_frame(U, cutoff_nm=CUTOFF_NM, use_bb_for_com=True):
    """
    Mutates positions of U.atoms in the current timestep:
      1) make residues in largest cluster contiguous (min image)
      2) translate cluster COM to box center (min image)
      3) wrap into primary box
    """
    box = U.dimensions.copy()
    anchors_all = _residue_anchors(U)

    cluster_res_idx = _largest_residue_cluster_from_anchors(anchors_all, cutoff_nm, box)

    idx = np.concatenate([U.residues[i].atoms.indices for i in cluster_res_idx])
    cluster = U.atoms[idx]
    com_group = cluster.select_atoms("name BB") if use_bb_for_com else cluster

    # 1) make residues contiguous near cluster COM
    com = com_group.center_of_mass()
    for ri in cluster_res_idx:
        res = U.residues[int(ri)]
        disp = res.atoms.center_of_mass() - com
        shift = minimize_vectors(disp[None, :], box=box)[0] - disp
        res.atoms.positions += shift

    # 2) translate cluster COM to box center
    com2 = com_group.center_of_mass()
    box_center = 0.5 * box[:3]
    delta = minimize_vectors((com2 - box_center)[None, :], box=box)[0]
    U.atoms.positions -= delta

    # 3) wrap
    U.atoms.pack_into_box(box=box)

def cell_vectors_from_dimensions(dim):
    """
    MDAnalysis dimensions: [a,b,c,alpha,beta,gamma]
      a,b,c in Å; angles in degrees.
    Return 3 cell vectors (Å) in Cartesian coordinates.
    """
    a, b, c, alpha, beta, gamma = dim
    alpha = math.radians(alpha)
    beta  = math.radians(beta)
    gamma = math.radians(gamma)

    va = np.array([a, 0.0, 0.0], dtype=float)
    vb = np.array([b * math.cos(gamma), b * math.sin(gamma), 0.0], dtype=float)

    cx = c * math.cos(beta)
    cy = c * (math.cos(alpha) - math.cos(beta) * math.cos(gamma)) / max(math.sin(gamma), 1e-8)
    cz_sq = c*c - cx*cx - cy*cy
    cz = math.sqrt(max(cz_sq, 0.0))
    vc = np.array([cx, cy, cz], dtype=float)

    return va, vb, vc

def unitcell_edges(va, vb, vc):
    """Edges for the parallelepiped cell."""
    o = np.array([0.0, 0.0, 0.0])
    pts = [
        o, va, vb, va+vb,
        vc, va+vc, vb+vc, va+vb+vc
    ]
    edges_idx = [
        (0,1),(0,2),(1,3),(2,3),
        (4,5),(4,6),(5,7),(6,7),
        (0,4),(1,5),(2,6),(3,7)
    ]
    return [(pts[i], pts[j]) for i, j in edges_idx]

def quat_axis_angle(axis, deg):
    ax, ay, az = axis
    n = (ax*ax + ay*ay + az*az)**0.5 or 1.0
    ax, ay, az = ax/n, ay/n, az/n
    th = math.radians(deg) * 0.5
    s = math.sin(th)
    return [ax*s, ay*s, az*s, math.cos(th)]

def quat_euler(degx=0.0, degy=0.0, degz=0.0, order="zyx"):
    def qmul(q2, q1):
        x1,y1,z1,w1 = q1
        x2,y2,z2,w2 = q2
        return [
            w2*x1 + x2*w1 + y2*z1 - z2*y1,
            w2*y1 - x2*z1 + y2*w1 + z2*x1,
            w2*z1 + x2*y1 - y2*x1 + z2*w1,
            w2*w1 - x2*x1 - y2*y1 - z2*z1
        ]
    qx = quat_axis_angle([1,0,0], degx)
    qy = quat_axis_angle([0,1,0], degy)
    qz = quat_axis_angle([0,0,1], degz)
    q  = [0,0,0,1]
    for c in order.lower():
        q = qmul({'x': qx, 'y': qy, 'z': qz}[c], q)
    return q

def quat_to_rotmat(q):
    x,y,z,w = q
    xx, yy, zz = x*x, y*y, z*z
    xy, xz, yz = x*y, x*z, y*z
    wx, wy, wz = w*x, w*y, w*z
    return np.array([
        [1-2*(yy+zz), 2*(xy-wz),   2*(xz+wy)],
        [2*(xy+wz),   1-2*(xx+zz), 2*(yz-wx)],
        [2*(xz-wy),   2*(yz+wx),   1-2*(xx+yy)]
    ], dtype=float)


def render_two_groups_pyvista(bb_pos, sc_pos, dims, out_png: Path, view_params,
                             bb_color=BB_COLOR, sc_color=SC_COLOR, line_color=PAPER_GRAY,
                             bb_point_size=BB_POINT_SIZE, sc_point_size=SC_POINT_SIZE,
                             window_size_px=WINDOW_SIZE_PX, dpi_scale=RENDER_DPI_SCALE):
    out_png = Path(out_png)
    out_png.parent.mkdir(parents=True, exist_ok=True)

    w, h = window_size_px
    w = int(w * dpi_scale)
    h = int(h * dpi_scale)

    pl = pv.Plotter(off_screen=False, window_size=(w, h))
    pl.set_background("white")
    pl.disable_anti_aliasing()
    
    if bb_pos is not None and len(bb_pos) > 0:
        bb = pv.PolyData(bb_pos)
        pl.add_mesh(bb, color=bb_color, opacity=0.98,
                    render_points_as_spheres=True, point_size=bb_point_size)

    if sc_pos is not None and len(sc_pos) > 0:
        sc = pv.PolyData(sc_pos)
        pl.add_mesh(sc, color=sc_color, opacity=0.9,
                    render_points_as_spheres=True, point_size=sc_point_size)

    va, vb, vc = cell_vectors_from_dimensions(dims)
    for p0, p1 in unitcell_edges(va, vb, vc):
        pl.add_mesh(pv.Line(p0, p1), color=line_color, line_width=2)

    pl.camera.parallel_projection = True

    if "cam_dir" in view_params:
        cam_dir = np.asarray(view_params["cam_dir"], float)
        cam_up  = np.asarray(view_params.get("cam_up", safe_up_for(cam_dir)), float)
    else:
        q = quat_euler(view_params["degx"], view_params["degy"], view_params["degz"], view_params["order"])
        R = quat_to_rotmat(q)
        cam_dir = R @ np.array([0.0, 0.0, 1.0])
        cam_up  = R @ np.array([0.0, 1.0, 0.0])


    box_center = 0.5 * (va + vb + vc)
    # Camera axes
    cam_dir = cam_dir / (np.linalg.norm(cam_dir) + 1e-12)
    cam_up  = cam_up  / (np.linalg.norm(cam_up)  + 1e-12)
    cam_right = np.cross(cam_dir, cam_up)
    cam_right = cam_right / (np.linalg.norm(cam_right) + 1e-12)

    # Focus on the aggregate (not the box)
    pts = []
    if bb_pos is not None and len(bb_pos) > 0: pts.append(bb_pos)
    if sc_pos is not None and len(sc_pos) > 0: pts.append(sc_pos)
    pts = np.vstack(pts) if len(pts) else np.zeros((1,3), dtype=float)

    focal = pts.mean(axis=0)

    # Orthographic scale: project onto camera plane and fit tightly
    rel = pts - focal
    x = rel @ cam_right
    y = rel @ cam_up
    half_span = float(max(np.max(np.abs(x)), np.max(np.abs(y)), 1.0))  # Å

    # pad factor: >1 means zoom OUT slightly; <1 means zoom IN
    pad = float(view_params.get("pad", 1.12))  # default: small margin
    pl.camera.parallel_scale = half_span * pad

    # Distance is irrelevant for ortho size, but keep it reasonable for clipping
    dist = 8.0 * half_span
    pl.camera.position = (focal + cam_dir * dist).tolist()
    pl.camera.focal_point = focal.tolist()
    pl.camera.up = cam_up.tolist()

    # Save without producing notebook output
    pl.show(screenshot=str(out_png), interactive=False, auto_close=True)

    if not is_valid_png(out_png):
        raise RuntimeError(f"Screenshot failed or invalid PNG: {out_png}")
    
def principal_axes(pts):
    """
    pts: (N,3) array in Å.
    Returns (axes, evals) where axes[:,0] is the longest-axis direction.
    """
    pts = np.asarray(pts, float)
    c = pts.mean(axis=0, keepdims=True)
    X = pts - c
    C = (X.T @ X) / max(len(X), 1)   # covariance-ish
    evals, evecs = np.linalg.eigh(C) # ascending
    order = np.argsort(evals)[::-1]  # descending
    evals = evals[order]
    evecs = evecs[:, order]          # columns are axes
    # normalize (just in case)
    evecs = evecs / (np.linalg.norm(evecs, axis=0, keepdims=True) + 1e-12)
    return evecs, evals

def safe_up_for(cam_dir, preferred=np.array([0.0, 1.0, 0.0])):
    cam_dir = np.asarray(cam_dir, float)
    cam_dir = cam_dir / (np.linalg.norm(cam_dir) + 1e-12)
    up = preferred.copy()
    # if nearly parallel, switch preferred
    if abs(np.dot(cam_dir, up)) > 0.92:
        up = np.array([1.0, 0.0, 0.0])
    # orthonormalize
    up = up - np.dot(up, cam_dir) * cam_dir
    up = up / (np.linalg.norm(up) + 1e-12)
    return up

def render_two_views_for_peptide(peptide_dir: Path, out_png1: Path, out_png2: Path, cutoff_nm=CUTOFF_NM):
    top = peptide_dir / TOP_NAME
    traj = peptide_dir / TRAJ_NAME
    if not top.exists() or not traj.exists():
        raise FileNotFoundError(f"Missing {TOP_NAME} or {TRAJ_NAME} in {peptide_dir}")

    u = mda.Universe(str(top), str(traj))
    u.trajectory[-1]  # last frame
    center_main_cluster_inplace_on_current_frame(u, cutoff_nm=cutoff_nm)

    dims = u.dimensions.copy()
    bb_pos = u.atoms.select_atoms("name BB").positions.copy()
    sc_pos = u.atoms.select_atoms("not name BB").positions.copy()

    pts = np.vstack([bb_pos, sc_pos]) if (len(bb_pos) and len(sc_pos)) else (bb_pos if len(bb_pos) else sc_pos)

    axes, evals = principal_axes(pts)
    a1 = axes[:, 0]
    a2 = axes[:, 1]
    a3 = axes[:, 2]

    anis = float(evals[0] / max(evals[1], 1e-12))

    if anis > 1.35:
        V1 = dict(cam_dir=a3, cam_up=safe_up_for(a3, preferred=a2), pad=VIEW_1.get("pad", 1.05))
        V2 = dict(cam_dir=a2, cam_up=safe_up_for(a2, preferred=a1), pad=VIEW_2.get("pad", 1.05))
    else:
        V1 = VIEW_1
        V2 = VIEW_2

    render_two_groups_pyvista(bb_pos, sc_pos, dims, out_png1, V1)
    render_two_groups_pyvista(bb_pos, sc_pos, dims, out_png2, V2)



def png_to_jpeg(png_path: Path, jpg_path: Path, quality=75, max_px=1400):
    """
    Convert to JPEG and optionally downscale so the largest side is max_px.
    """
    png_path = Path(png_path)
    jpg_path = Path(jpg_path)
    jpg_path.parent.mkdir(parents=True, exist_ok=True)

    im = Image.open(png_path).convert("RGB")
    w, h = im.size
    m = max(w, h)
    if m > max_px:
        scale = max_px / m
        im = im.resize((int(w*scale), int(h*scale)), Image.Resampling.LANCZOS)

    im.save(jpg_path, "JPEG", quality=quality, optimize=True, progressive=True)
    return jpg_path

def build_group_pdf_onepage(entries, out_pdf: Path, title: str, subtitle: str = "",
                            max_cols: int = 3,
                            margin_cm: float = 0.7,
                            gutter_cm: float = 0.28,
                            inner_gutter_cm: float = 0.18,
                            caption_fs: int = 8,
                            header_fs: int = 12,
                            subtitle_fs: int = 8):
    """
    Single-page portrait contact sheet:
      - grid of peptides (<= max_cols columns)
      - each tile: caption + two views
    """
    out_pdf = Path(out_pdf)
    out_pdf.parent.mkdir(parents=True, exist_ok=True)

    page_w, page_h = A4
    page_h = page_h - 3*cm
    c = canvas.Canvas(str(out_pdf), pagesize=(page_w, page_h))

    margin = margin_cm * cm
    gutter = gutter_cm * cm
    inner_gutter = inner_gutter_cm * cm

    # Header height (kept compact)
    header_h = 0.8 * cm if (title or subtitle) else 0.0

    usable_w = page_w - 2 * margin
    usable_h = page_h - 2 * margin - header_h

    n = len(entries)
    if n == 0:
        c.setFont("Helvetica-Bold", header_fs)
        c.drawString(margin, page_h - margin, title or "Empty group")
        c.setFont("Helvetica", subtitle_fs)
        c.drawString(margin, page_h - margin - 0.8*cm, "No peptides found.")
        c.save()
        return

    # Fixed max 3 columns; choose cols based on count
    cols = min(max_cols, 3)
    if n <= 2: cols = n
    elif n <= 6: cols = min(2, cols)

    rows = int(ceil(n / cols))

    tile_w = (usable_w - (cols - 1) * gutter) / cols
    tile_h = (usable_h - (rows - 1) * gutter) / rows

    # Caption band
    cap_h = 0.55 * cm  # bigger than before

    # Header
    y_top = page_h - margin
    if header_h > 0:
        c.setFont("Helvetica-Bold", header_fs)
        c.drawString(margin, y_top, title)
        if subtitle:
            c.setFont("Helvetica", subtitle_fs)
            c.drawString(margin, y_top - 0.60*cm, subtitle)

    grid_top = page_h - margin - header_h

    def draw_placeholder(x, y, w, h, text="missing"):
        c.rect(x, y, w, h, stroke=1, fill=0)
        c.setFont("Helvetica", max(7, caption_fs - 1))
        c.drawString(x + 0.15*cm, y + h - 0.45*cm, text)

    # Sort so the grid is meaningful (AP desc, then RMOI desc)
    def sort_key(e):
        ap = e.get("ap"); rm = e.get("rmoi")
        apk = ap if ap is not None else -1e18
        rmk = rm if rm is not None else -1e18
        return (-apk, -rmk, str(e.get("peptide","")))

    entries_sorted = sorted(entries, key=sort_key)

    for idx, e in enumerate(entries_sorted):
        r = idx // cols
        col = idx % cols

        tx = margin + col * (tile_w + gutter)
        ty_top = grid_top - r * (tile_h + gutter)
        ty = ty_top - tile_h

        # Caption (larger)
        peptide = str(e.get("peptide", "NA"))
        ap = e.get("ap"); rm = e.get("rmoi")
        cap = f"{peptide} | AP: {(f'{ap:.3f}' if ap is not None else 'NA')} | RMOI: {(f'{rm:.3f}' if rm is not None else 'NA')}"
        c.setFont("Helvetica", caption_fs)
        c.drawString(tx + 0.05*cm, ty_top - 0.45*cm, cap[:120])

        # Image area (use as much of tile as possible)
        img_area_h = tile_h - cap_h
        img_w = (tile_w - inner_gutter) / 2.0
        img_h = img_area_h

        img_y = ty
        img1 = Path(e.get("img1_path") or "")
        img2 = Path(e.get("img2_path") or "")

        def draw_img(path: Path, x, y, w, h):
            if (not path.exists()) or (path.stat().st_size < MIN_PNG_BYTES):
                draw_placeholder(x, y, w, h, "render failed")
                return
            try:
                jpg_path = Path(str(path)).with_suffix(".jpg")
                if (not jpg_path.exists()) or jpg_path.stat().st_mtime < path.stat().st_mtime:
                    png_to_jpeg(path, jpg_path, quality=75, max_px=1200)
                ir = ImageReader(str(jpg_path))
                # Fill the allocated box (minimal letterboxing)
                c.drawImage(ir, x, y, width=w, height=h, preserveAspectRatio=True, anchor='c', mask="auto")
            except Exception:
                draw_placeholder(x, y, w, h, "bad image")

        draw_img(img1, tx,                    img_y, img_w, img_h)
        draw_img(img2, tx + img_w + inner_gutter, img_y, img_w, img_h)

        # Tile outline (optional but helps scanning)
        c.setLineWidth(0.4)
        c.rect(tx, ty, tile_w, tile_h, stroke=1, fill=0)

    c.save()
```

```python
merged_by_run = pd.read_csv(CSV_PATH)

for col in ["peptide", "class", "run", "aggregation_propensity", "RMOI"]:
    if col not in merged_by_run.columns:
        raise ValueError(f"CSV missing required column: {col}")

merged_by_run["aggregation_propensity"] = pd.to_numeric(merged_by_run["aggregation_propensity"], errors="coerce")
merged_by_run["RMOI"] = pd.to_numeric(merged_by_run["RMOI"], errors="coerce")

if CLASSES is None:
    CLASSES = sorted(merged_by_run["class"].dropna().unique().tolist())

print("Runs:", RUNS)
print("Classes:", CLASSES)
```

```python
OUT_ROOT.mkdir(parents=True, exist_ok=True)

def peptide_dir_for(run: str, cls: str, peptide: str) -> Path:
    return ROOT / run / cls / peptide

def cache_paths(peptide_dir: Path):
    cache_dir = peptide_dir / IMG_CACHE_DIRNAME
    return cache_dir / "view_1.png", cache_dir / "view_2.png"

def ensure_two_views(peptide_dir: Path, force=False):
    img1, img2 = cache_paths(peptide_dir)

    if (not force) and is_valid_png(img1) and is_valid_png(img2):
        return img1, img2

    # Always (re)create cache dir
    img1.parent.mkdir(parents=True, exist_ok=True)

    # Render deterministically with PyVista
    render_two_views_for_peptide(peptide_dir, img1, img2, cutoff_nm=CUTOFF_NM)
    return img1, img2

def get_peptides_for_group(run, cls):
    if USE_FILESYSTEM:
        base = ROOT / run / cls
        if not base.exists():
            return []
        return sorted([p.name for p in base.iterdir() if p.is_dir()])
    else:
        df = merged_by_run[(merged_by_run["run"] == run) & (merged_by_run["class"] == cls)]
        return df["peptide"].dropna().astype(str).unique().tolist()
```

```python
summary = []

for run in RUNS:
    for cls in CLASSES:
        df = merged_by_run[(merged_by_run["run"] == run) & (merged_by_run["class"] == cls)].copy()
        peptides = get_peptides_for_group(run, cls)

        entries = []
        print(f"\n=== {run} | {cls} | peptides={len(peptides)} ===")

        for pep in peptides:
            pdir = peptide_dir_for(run, cls, pep)

            ap = None
            rm = None
            if not df.empty and (df["peptide"] == pep).any():
                row = df.loc[df["peptide"] == pep].iloc[0]
                ap = float(row["aggregation_propensity"]) if pd.notna(row["aggregation_propensity"]) else None
                rm = float(row["RMOI"]) if pd.notna(row["RMOI"]) else None

            img1 = img2 = None
            try:
                if pdir.exists():
                    img1, img2 = ensure_two_views(pdir, force=FORCE_RERENDER)
                else:
                    print(f"  [WARN] Missing directory: {pdir}")
            except Exception as ex:
                print(f"  [WARN] Render failed for {pep} ({pdir}): {ex}")

            entries.append(dict(
                peptide=pep,
                ap=ap,
                rmoi=rm,
                img1_path=str(img1) if img1 else None,
                img2_path=str(img2) if img2 else None,
            ))

        section = SECTION_LABEL.get(cls, cls)
        cohort  = COHORT_LABEL.get(cls, cls)

        safe_name = cls
        out_pdf = OUT_ROOT / run / f"{safe_name}.pdf"

        build_group_pdf_onepage(
            entries,
            out_pdf,
            title=f"{run} | {section} | {cohort}",
            # subtitle="Two orthographic views; last frame; centered main cluster",
            max_cols=3,
            caption_fs=8,
            header_fs =12,
            subtitle_fs= 8,
        )
        
        summary.append((run, cls, len(entries), str(out_pdf)))

print("\nDone. PDFs written:")
for run, cls, n, path in summary:
    print(f"  {run:>5} | {cls:<20} | peptides={n:4d} | {path}")
```
