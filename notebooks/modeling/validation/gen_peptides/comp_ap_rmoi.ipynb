```python
import pandas as pd
import numpy as np

# load negative control structures
neg_ctrl_df = pd.read_csv('negative_control_structures.csv')
neg_ctrl_df
```

```python
# rename column 'structure_id' to 'peptide'
neg_ctrl_df = neg_ctrl_df.rename(columns={'structure_id': 'peptide'})
```

```python
# load rmoi_and_ap_summary_total
rmoi_ap_summary_df = pd.read_csv('rmoi_and_ap_summary_total.csv')
rmoi_ap_summary_df
```

```python
# left join on neg_ctrl_df and rmoi_ap_summary_df on 'structure_id' and 'peptide'
merged_df = pd.merge(neg_ctrl_df, rmoi_ap_summary_df, on=['peptide'], how='left')
merged_df
```

```python
# get df with random peptides, where structure_type is 'random'
random_peptides_df = merged_df[merged_df['structure_type'] == 'random']
random_peptides_df
```

```python
# check how many have aggregation_propensity_mean > 1.8
high_ap_random_df = random_peptides_df[random_peptides_df['aggregation_propensity_mean'] > 1.8]
high_ap_random_df
```

```python
# get df with sphere and fiber structure_type, which have ap as NaN
sphere_fiber_pre_ap_df = merged_df[(merged_df['structure_type'].isin(['sphere', 'fiber'])) & (merged_df['ap'].isna())]
sphere_fiber_pre_ap_df
```

```python
# check how many have > 1.8
high_rmoi_sphere_fiber_df = sphere_fiber_pre_ap_df[sphere_fiber_pre_ap_df['aggregation_propensity_mean'] > 1.8]
high_rmoi_sphere_fiber_df
```

```python
# get df with sphere and fiber structure_type, which have ap as not NaN
sphere_fiber_post_ap_df = merged_df[(merged_df['structure_type'].isin(['sphere', 'fiber'])) & (~merged_df['ap'].isna())]
sphere_fiber_post_ap_df
```

```python
# check how many have ap > 1.8
high_ap_sphere_fiber_df = sphere_fiber_post_ap_df[sphere_fiber_post_ap_df['aggregation_propensity_mean'] > 1.8]
high_ap_sphere_fiber_df
```

```python
# get df with unsupervised type
unsupervised_df = rmoi_ap_summary_df[rmoi_ap_summary_df['morphology'] == 'unsupervised']
unsupervised_df
```

```python
# check how many have aggregation_propensity_mean > 1.8
high_ap_unsupervised_df = unsupervised_df[unsupervised_df['aggregation_propensity_mean'] > 1.8]
high_ap_unsupervised_df
```
